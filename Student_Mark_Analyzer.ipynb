# Student Mark Analyzer
This notebook calculates totals, percentages, grades, pass status, class average, topper, and subject averages.

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv('student_marks.csv')
df.head()
```

## Calculate totals, percentages, grades, and result

```python
subjects = [c for c in df.columns if c not in ['Student_ID', 'Student_Name']]
df['Total'] = df[subjects].sum(axis=1)
df['Percentage'] = (df['Total'] / (len(subjects) * 100) * 100).round(2)
df['Result'] = df[subjects].ge(35).all(axis=1).map({True: 'PASS', False: 'NEEDS IMPROVEMENT'})
df['Grade'] = pd.cut(df['Percentage'], bins=[-1,39.99,49.99,59.99,69.99,79.99,100], labels=['F','D','C','B','A','A+'])
df
```

## Class insights

```python
print('Class average:', round(df['Percentage'].mean(), 2), '%')
topper = df.loc[df['Percentage'].idxmax()]
print('Topper:', topper['Student_Name'], '-', topper['Percentage'], '%')
print('\nSubject averages:')
print(df[subjects].mean().round(2))
```

## Student performance chart

```python
plt.figure(figsize=(9,5))
plt.bar(df['Student_Name'], df['Percentage'])
plt.title('Student-wise Percentage')
plt.xlabel('Student')
plt.ylabel('Percentage')
plt.ylim(0, 100)
plt.xticks(rotation=30)
plt.tight_layout()
plt.show()
```

## Export results

```python
df.to_csv('student_analysis_output.csv', index=False)
print('Analysis saved to student_analysis_output.csv')
```
